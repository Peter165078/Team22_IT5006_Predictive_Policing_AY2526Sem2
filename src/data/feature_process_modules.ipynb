```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

pd.set_option("display.max_columns", None)

def get_project_root() -> Path:
    current = Path(__file__).resolve() if "__file__" in globals() else Path().resolve()

    for parent in [current] + list(current.parents):
        if (parent / ".git").exists():
            return parent

    raise RuntimeError("Project root not found. Ensure you're inside a Git project.")

# ---- Core paths ----
PROJECT_ROOT = get_project_root()

DATA_PATH = PROJECT_ROOT / "data" / "raw"
OUTPUT_DIR = PROJECT_ROOT / "artifacts" / "figures"

# Ensure output directory exists
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
df = pd.read_csv(f"{DATA_PATH}/Crimes_-_2001_to_Present_20260216.csv")
df.shape
```

```python
df["Date"] = pd.to_datetime(df["Date"],
                            format="%m/%d/%Y %I:%M:%S %p",
                            errors="coerce")
```

```python
# 筛选 2015-2025 年的数据
mask = (df["Date"].dt.year >= 2022) & (df["Date"].dt.year <= 2025)
df_filtered = df.loc[mask].copy()

print(f"筛选前: {df.shape}")
print(f"筛选后: {df_filtered.shape}")
print(f"年份分布:\n{df_filtered['Date'].dt.year.value_counts().sort_index()}")

# 保存新 CSV
output_path = f"{DATA_PATH}/Crimes_2022_to_2025.csv"
df_filtered.to_csv(output_path, index=False)
print(f"\n已保存至: {output_path}")
```
